## Import das bibliotecas necessárias

```python
import os
import base64
from dotenv import load_dotenv
from openai import AzureOpenAI
from openai import OpenAI # for Grok

load_dotenv()
print("Environment variables loaded.")
endpoint = os.getenv("AZURE_OPENAI_ENDPOINT_APHAX")
print(f"Endpoint retrieved: {endpoint}")
```

```python
GPT_5_MINI = "gpt-5-mini"
GPT_4o = "gpt-4o"
GROK_3 = "grok-3"
TEXT_EMBEDDING_3_LARGE = "text-embedding-3-large"


def create_client(endpoint: str, api_key: str, api_version: str) -> AzureOpenAI:
     return AzureOpenAI(
        api_version=api_version,
        azure_endpoint=endpoint,
        api_key=api_key
    )

def generate_response(client: OpenAI, model: str, system_behavior: str, prompt: str) -> str:
    messages = [{"role": "system", "content": system_behavior},{"role": "user", "content": prompt}]

    response = client.chat.completions.create(
        model=model,
        messages=messages
    )
    return response.choices[0].message.content


def generate_embedding(client: OpenAI, model: str, input_text: str) -> list:
    response = client.embeddings.create(
        model=model,
        input=input_text
    )
    return response.data[0].embedding
```

```python
client = AzureOpenAI(
    api_version="2024-12-01-preview",
    azure_endpoint=os.getenv("AZURE_OPENAI_ENDPOINT_APHAX"),
    api_key=os.getenv("AZURE_OPENAI_API_KEY_ALPHAX")
)
```

```python
system_behavior = " Você um agente que responde de forma amigável e sempre em portugês do Brasil."

prompt = "Explique a teoria da relatividade de forma simples."

response = generate_response(client, GPT_5_MINI, system_behavior, prompt)
print("Response from GPT-5 Mini:")
print(response)
```

```python
GROK_ENDPOINT = os.getenv("AZURE_GROK_ENDPOINT")
GROK_KEY = os.getenv("AZURE_GROK_KEY")

client_grok = OpenAI(
    base_url=GROK_ENDPOINT,
    api_key=GROK_KEY
)

r = generate_response(client=client_grok,
    model=GROK_3,
    system_behavior=system_behavior,
    prompt=prompt 
)
print(r)
```
